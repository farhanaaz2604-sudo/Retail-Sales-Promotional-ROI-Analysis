```python
import pandas as pd
# Define the age boundaries and corresponding labels
bins = [17, 25, 40, 55, 100]
labels = ['18-25 (Gen Z)', '26-40 (Millennials)', '41-55 (Gen X)', '56+ (Boomers)']

# Create the new age_group column
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
```

```python
df[["age","age_group"]].head(10)
```

```python
import numpy as np
import pandas as pd

# Define the business conditions in order of priority
conditions = [
    # 1. High-Value Loyal: Spends $60+ and has 20+ previous purchases
    (df['purchase amount (usd)'] >= 60) & (df['previous purchases'] >= 20),
    
    # 2. Discount-Driven: Used a discount and spends below $60
    (df['discount applied'] == 'Yes') & (df['purchase amount (usd)'] < 60),
    
    # 3. At-Risk: Low repeat history (<= 10 previous purchases) and no active subscription
    (df['previous purchases'] <= 10) & (df['subscription status'] == 'No')
]

# Define the matching labels
tiers = [
    'High-Value Loyal',
    'Discount-Driven',
    'At-Risk'
]
# Create the column (customers who don't fit the above fall into 'Standard Regular')
df['customer_value_tier'] = np.select(conditions, tiers, default='Standard Regular')
```

```python
# Verify the results
print(df[['customer id', 'purchase amount (usd)', 'previous purchases', 'discount applied', 'customer_value_tier']].head(10))
print("\nCustomer count per Value Tier:")
print(df['customer_value_tier'].value_counts())
```

```python
# Creating frequency of purchase
frequency_mapping = {
    "Fortnightly":14,
    "Weekly": 7,
    "Monthly": 30,
    "Quarterly": 90,
    "Bi-Weekly":14,
    "Annually":365,
    "Every 3 Months":90
}
df["purchase_frequency_days"] = df["frequency of purchases"].map(frequency_mapping)
```
